# 01 — EDA and classical baseline

This notebook loads the Banking77 dataset, explores the label distribution, and trains a TF-IDF + logistic regression baseline.

```python
from src.intent.data import load_raw
from src.intent.models import train_baseline
import matplotlib.pyplot as plt
import seaborn as sns

train, test = load_raw()
print(train.shape, test.shape)
```

```python
# Class distribution
plt.figure(figsize=(12, 6))
order = train['label'].value_counts().index
sns.countplot(data=train, y='label', order=order)
plt.title('Training set intent distribution')
plt.tight_layout()
plt.show()
```

```python
# Train baseline
clf, vectorizer, le, metrics = train_baseline(train, test)
print(metrics['accuracy'], metrics['macro_f1'])
```
